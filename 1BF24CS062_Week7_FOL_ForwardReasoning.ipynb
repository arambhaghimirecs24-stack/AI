```python
def solve_forward_chaining():
    # 1. Define initial facts (Known truths)
    # We represent facts as strings
    facts = {
        "American(Robert)",
        "Enemy(Country_A, America)",
        "Owns(Country_A, Missiles)",
        "Missile(Missiles)"
    }

    # 2. Define the Rules as a list of dictionaries with premises and conclusion
    # Rule 1: All missiles are weapons.
    # Rule 2: If Robert sold all missiles to Country A, then Robert sold Missiles to Country A.
    # Rule 3: An enemy of America is a hostile nation.
    # Rule 4: If an American sells weapons to a hostile nation, they are a criminal.
    rules = [
        {
            "name": "Rule 1 (Missiles are weapons)",
            "apply": lambda f:
                {"Weapon(Missiles)"} if "Missile(Missiles)" in f else set()
        },
        {
            "name": "Rule 2 (Robert sold weapons to Country A)",
            "apply": lambda f:
                {"Sells(Robert, Missiles, Country_A)"} if "Owns(Country_A, Missiles)" in f else set()
        },
        {
            "name": "Rule 3 (Enemy nation is hostile)",
            "apply": lambda f:
                {"Hostile(Country_A)"} if "Enemy(Country_A, America)" in f else set()
        },
        {
            "name": "Rule 4 (Crime Rule)",
            "apply": lambda f:
                {"Criminal(Robert)"} if ("American(Robert)" in f and
                                       "Weapon(Missiles)" in f and
                                       "Sells(Robert, Missiles, Country_A)" in f and
                                       "Hostile(Country_A)" in f) else set()
        }
    ]

    print("--- Starting Forward Chaining Process ---")
    print(f"Initial Facts: {sorted(list(facts))}\n")

    # Forward chaining loop
    step = 1
    new_facts_inferred = True

    while new_facts_inferred:
        new_facts_inferred = False
        inferred_this_round = set()

        for rule in rules:
            new_facts = rule["apply"](facts)
            # Check if we derived any new facts not already in our database
            for fact in new_facts:
                if fact not in facts:
                    inferred_this_round.add(fact)
                    new_facts_inferred = True

        if new_facts_inferred:
            print(f"Step {step}: Inferred {inferred_this_round}")
            facts.update(inferred_this_round)
            print(f"Current Knowledge Base: {sorted(list(facts))}\n")
            step += 1

            if "Criminal(Robert)" in facts:
                print("Goal 'Criminal(Robert)' has been proven!")
                break
        else:
            print("No more facts can be inferred.")

    print("\n--- Final Result ---")
    if "Criminal(Robert)" in facts:
        print("PROVEN: Robert is indeed a criminal.")
    else:
        print("FAILED: Could not prove that Robert is a criminal.")

solve_forward_chaining()
```
