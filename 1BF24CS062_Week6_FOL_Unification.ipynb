```python
# Unification in First Order Logic (FOL)

substitution = {}

def is_variable(x):
    return x.islower() and x.isalpha()

def apply_substitution(term):
    while term in substitution:
        term = substitution[term]
    return term

def unify(x, y):
    x = apply_substitution(x)
    y = apply_substitution(y)

    if x == y:
        return True

    if is_variable(x):
        substitution[x] = y
        return True

    if is_variable(y):
        substitution[y] = x
        return True

    return False


def main():
    print("Expression 1: Eats(x, Apple)")
    print("Expression 2: Eats(Riya, y)")

    # Unify the corresponding arguments
    result1 = unify("x", "Riya")
    result2 = unify("Apple", "y")

    if result1 and result2:
        print("\nUnification Successful!")
        print("Substitutions:")

        for var, value in substitution.items():
            print(f"{var} = {value}")

        print("\nUnified Expression: Eats(Riya, Apple)")
    else:
        print("\nUnification Failed!")


if __name__ == "__main__":
    main()
```
