```python
import math

class AlphaBetaTreeVisualizer:
    def __init__(self):
        self.pruned_nodes = set()
        self.visited_nodes = set()

    def run_demo(self):
        # Construct a classic game tree of depth 3 to show actual pruning
        # Root (Max) -> Left/Right (Min) -> Sub-children (Max) -> Leaf Nodes
        #
        #                  Root (MAX)
        #                /            \
        #             A (MIN)          B (MIN)
        #            /       \        /       \
        #          C (MAX)  D (MAX)  E (MAX)  F (MAX)
        #         /   \    /   \    /   \    /   \
        #        3     5  6     9  1     2  0    -1

        tree = {
            'Root': ['A', 'B'],
            'A': ['C', 'D'],
            'B': ['E', 'F'],
            'C': [3, 5],
            'D': [6, 9],
            'E': [1, 2],
            'F': [0, -1]
        }

        print("\n--- Running Alpha-Beta Search & Tracking Pruned Branches ---\n")
        optimal_val = self.alpha_beta(tree, 'Root', 0, -math.inf, math.inf, True)

        print(f"\nOptimal Value at Root: {optimal_val}\n")
        self.draw_tree(tree, 'Root', 0)

    def alpha_beta(self, tree, node, depth, alpha, beta, is_max):
        self.visited_nodes.add(node)

        # If it's a leaf node (numeric value)
        if isinstance(node, (int, float)):
            return node

        children = tree.get(node, [])

        if is_max:
            v = -math.inf
            for child in children:
                v = max(v, self.alpha_beta(tree, child, depth + 1, alpha, beta, False))
                alpha = max(alpha, v)
                if beta <= alpha:
                    # Mark remaining children of this node as pruned
                    idx = children.index(child)
                    for pruned_child in children[idx+1:]:
                        self.mark_pruned_subtree(tree, pruned_child)
                    break
            return v
        else:
            v = math.inf
            for child in children:
                v = min(v, self.alpha_beta(tree, child, depth + 1, alpha, beta, True))
                beta = min(beta, v)
                if beta <= alpha:
                    # Mark remaining children of this node as pruned
                    idx = children.index(child)
                    for pruned_child in children[idx+1:]:
                        self.mark_pruned_subtree(tree, pruned_child)
                    break
            return v

    def mark_pruned_subtree(self, tree, node):
        self.pruned_nodes.add(node)
        if not isinstance(node, (int, float)):
            for child in tree.get(node, []):
                self.mark_pruned_subtree(tree, child)

    def draw_tree(self, tree, node, depth, prefix=""):
        if node in self.pruned_nodes:
            status = "[PRUNED]"
        elif node in self.visited_nodes:
            status = "[VISITED]"
        else:
            status = ""

        indent = "    " * depth
        print(f"{indent}└── {node} {status}")

        if not isinstance(node, (int, float)):
            for child in tree.get(node, []):
                self.draw_tree(tree, child, depth + 1)

visualizer = AlphaBetaTreeVisualizer()
visualizer.run_demo()
```
